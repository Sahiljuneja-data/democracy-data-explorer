```python
india_1900 = df[(df['year'] >= 1900) & (df['country_name'] == 'India')]
```

```python
india_1900[["country_name", "year", "v2x_libdem"]].head()
```

```python
india_1900["v2x_libdem"].describe()
```

```python
import matplotlib.pyplot as plt
plt.plot(india_1900["year"], india_1900["v2x_libdem"])
plt.xlabel("year")
plt.ylabel("v2x_libdem")
plt.title("lib dem in India 1900-2025")

plt.show()
```

```python
pakistan_1900 = df[(df['country_name'] == "Pakistan") & (df["year"] >=1900)]
```

```python
pakistan_1900[["country_name", "year", "v2x_libdem"]].head()
```

```python
pakistan_1900["v2x_libdem"].describe()
```

```python
plt.plot(pakistan_1900["year"], pakistan_1900["v2x_libdem"])
plt.xlabel("year")
plt.ylabel("v2x_libdem")
plt.title("lib dem in Pakistan 1900-2025")
```

```python
plt.plot(india_1900["year"], india_1900["v2x_libdem"], label="India")
plt.plot(pakistan_1900["year"], pakistan_1900["v2x_libdem"], label="Pakistan")

plt.xlabel("Year")
plt.ylabel("Liberal Democracy Index")
plt.title("Liberal Democracy in India and Pakistan")

plt.legend()

plt.show()
```
